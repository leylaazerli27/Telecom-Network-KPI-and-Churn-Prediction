```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = "https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv"
```

```python
df = pd.read_csv(url)
```

```python
print("Dataset shape (rows, columns):", df.shape)
```

```python
display(df.head())
```

```python
print("\nDataset info (data types and non-null counts):")
df.info()
```

```python
duplicates = df['customerID'].duplicated().sum()
print(f"Duplicate customer IDs: {duplicates}")
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
missing_total_charges = df['TotalCharges'].isnull().sum()
print(f"Missing values in TotalCharges: {missing_total_charges}")
```

```python
display(df[df['TotalCharges'].isnull()][['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges']])
```

```python
df['TotalCharges'] = df['TotalCharges'].fillna(0.0)
```

```python
print("\nUpdated TotalCharges data type:", df['TotalCharges'].dtype)
print("Remaining missing values in dataset:", df.isnull().sum().sum())
```

```python
np.random.seed(42)
n = len(df)
```

```python
base_load = np.where(
    df['Churn'] == 'Yes',
    np.random.normal(loc=82, scale=10, size=n),
    np.random.normal(loc=64, scale=14, size=n)
)
df['Base_Station_Load_Pct'] = np.clip(base_load, 25, 99).round(1)
```

```python
# Good signal: -75 to -90 dBm | Poor signal: -105 to -120 dBm
rsrp = np.where(
    df['Churn'] == 'Yes',
    np.random.normal(loc=-103, scale=8, size=n),
    np.random.normal(loc=-91, scale=9, size=n)
)
df['Avg_RSRP_dBm'] = np.clip(rsrp, -122, -70).round(1)
# Worse signal (RSRP < -100 dBm) increases the probability of dropped calls
drop_lambda = np.where(df['Avg_RSRP_dBm'] < -100, 3.2, 0.7)
df['Dropped_Calls_Month'] = np.random.poisson(lam=drop_lambda)
```

```python
speed = np.zeros(n)
for i in range(n):
    service = df.loc[i, 'InternetService']
    load_factor = 1 - (df.loc[i, 'Base_Station_Load_Pct'] / 130.0) # High load reduces throughput
    if service == 'Fiber optic':
        speed[i] = np.random.normal(loc=180, scale=35) * load_factor
    elif service == 'DSL':
        speed[i] = np.random.normal(loc=45, scale=10) * load_factor
    else:
        speed[i] = 0.0 # No internet service

df['Avg_Download_Mbps'] = np.clip(speed, 0, 300).round(1)        
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score

# 1. Prepare features (X) and target variable (y)
# Drop customerID as it is just an identifier
X = df.drop(columns=['customerID', 'Churn'])
y = (df['Churn'] == 'Yes').astype(int)

# Convert categorical text columns into numeric dummy variables (One-Hot Encoding)
X_encoded = pd.get_dummies(X, drop_first=True)

# 2. Split dataset into Training (80%) and Testing (20%) sets
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y
)

# 3. Train Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# 4. Evaluate model performance on unseen test data
y_pred = rf_model.predict(X_test)
y_prob = rf_model.predict_proba(X_test)[:, 1]

print(f"Model ROC-AUC Score: {roc_auc_score(y_test, y_prob):.3f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# 5. Extract and plot Top-10 Most Important Features driving Customer Churn
importances = pd.Series(rf_model.feature_importances_, index=X_encoded.columns)
top_10_features = importances.sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=top_10_features.values, y=top_10_features.index, color="#2b5c8f")
plt.title("Top-10 Key Drivers of Telecom Customer Churn (Random Forest)", fontsize=13, fontweight="bold")
plt.xlabel("Feature Importance Score", fontsize=11)
plt.ylabel("Feature", fontsize=11)
plt.tight_layout()
plt.savefig("feature_importance.png", dpi=300)
plt.show()
```

```python
import sqlite3

# 1. Create an in-memory SQLite database and load our clean DataFrame as a SQL table
conn = sqlite3.connect(":memory:")
df.to_sql("telecom_customers", conn, index=False, if_exists="replace")

# 2. SQL Query 1: Segment customers by Signal Quality (RSRP) and calculate Churn Rate & Revenue Loss
query_signal_zones = """
SELECT 
    CASE 
        WHEN Avg_RSRP_dBm >= -90 THEN '1. Good Signal (>= -90 dBm)'
        WHEN Avg_RSRP_dBm >= -100 THEN '2. Fair Signal (-90 to -100 dBm)'
        ELSE '3. Poor Signal (< -100 dBm)'
    END AS Signal_Zone,
    COUNT(customerID) AS Total_Customers,
    SUM(CASE WHEN Churn = 'Yes' THEN 1 ELSE 0 END) AS Churned_Customers,
    ROUND(AVG(CASE WHEN Churn = 'Yes' THEN 1.0 ELSE 0.0 END) * 100, 2) AS Churn_Rate_Pct,
    ROUND(AVG(Dropped_Calls_Month), 2) AS Avg_Dropped_Calls,
    ROUND(SUM(CASE WHEN Churn = 'Yes' THEN MonthlyCharges ELSE 0 END), 2) AS Monthly_Revenue_Lost_USD
FROM telecom_customers
GROUP BY Signal_Zone
ORDER BY Signal_Zone;
"""

sql_result_1 = pd.read_sql_query(query_signal_zones, conn)
print("SQL Query 1: Network Signal Quality vs. Churn & Revenue Loss")
display(sql_result_1)

# 3. SQL Query 2: Identify High-Risk VIP Customers (High Monthly Charges + Poor Network + Active Status)
# These are current customers we must save immediately!
query_vip_risk = """
SELECT 
    customerID,
    Contract,
    InternetService,
    MonthlyCharges,
    Avg_RSRP_dBm,
    Base_Station_Load_Pct,
    Dropped_Calls_Month
FROM telecom_customers
WHERE Churn = 'No'
  AND MonthlyCharges > 80
  AND Avg_RSRP_dBm < -102
  AND Base_Station_Load_Pct > 80
ORDER BY MonthlyCharges DESC
LIMIT 10;
"""

sql_result_2 = pd.read_sql_query(query_vip_risk, conn)
print("\nSQL Query 2: Top-10 High-Value Active Customers at Critical Technical Risk")
display(sql_result_2)
```
